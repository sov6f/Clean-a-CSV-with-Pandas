```python
categorical_cols = [
    "Employee_ID",
    "First_Name",
    "Last_Name",
    "Department_Region",
    "Status",
    "Join_Date",
    "Email",
    "Performance_Score"
]

for col in categorical_cols:
    print(f"\n--- {col} ({df[col].nunique()} unique values) ---")
    print(sorted(df[col].dropna().unique(), key=str))
```
